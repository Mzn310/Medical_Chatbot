```python
from langchain_community.document_loaders import (
    PyPDFLoader,
    DirectoryLoader,
)
```

```python
def load_pdf_files(data):
    loader = DirectoryLoader(data,glob="*.pdf",loader_cls=PyPDFLoader)
    documents=loader.load()
    return documents
    
```
